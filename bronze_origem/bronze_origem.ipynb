```sql
%python
#consultando os arquivos originais
display(dbutils.fs.ls('/Volumes/workspace/techvenda/filestore/origem/'))
```

```sql
%python
#criando variaveis para armazenar os caminhos
bronze_path = '/Volumes/workspace/techvenda/filestore/bronze/'
silver_path = '/Volumes/workspace/techvenda/filestore/silver/'
gold_path = '/Volumes/workspace/techvenda/filestore/gold/'
origem_path = '/Volumes/workspace/techvenda/filestore/origem/'
```

```sql
%python
#consulta dos arquivos na variavel bronze_path
display(dbutils.fs.ls(origem_path))
```

```sql
%python
# ler os arquivos originais do caminho especificado, identifica a primeira linha como cabeçalho e armazena no Dataframe
df_clientes = spark.read.csv(f'{origem_path}/clientes.csv',
                           header=True,
                           inferSchema=True,
                           sep=','
                           
                           )
df_itens_pedido = spark.read.csv(f'{origem_path}/itens_pedido.csv',
                                 header=True,
                                 inferSchema=True,
                                 sep=','
)
df_pedidos = spark.read.csv(f'{origem_path}/pedidos.csv',
                            header=True,
                            inferSchema=True,
                            sep=','
)
df_produtos = spark.read.csv(f'{origem_path}/produtos.csv',
                             header=True,
                             inferSchema=True,
                             sep=','
)
df_vendedores = spark.read.csv(f'{origem_path}/vendedores.csv',
                               header=True,
                               inferSchema=True,
                               sep=','
)       
```

Consultando todas os arquivos em tabelas

```sql
%python
display(df_clientes)
```

```sql
%python
display(df_itens_pedido)
```

```sql
%python
#identificado uma coluna com descrição errada, renomeando a coluna
df_itens_pedido = df_itens_pedido.withColumnRenamed("preco_unitario","marca")
```

```sql
%python
display(df_itens_pedido)
```

```sql
%python
#Salvando o DataFrame das tabelas como uma tabela Delta na camada Bronze, sobrescrevendo dados existentes e permitindo alterações no schema.
df_clientes.write\
    .mode('overwrite')\
        .format('delta')\
            .option('mergeSchema', 'true')\
                .save(f'{bronze_path}/clientes')
                
df_itens_pedido.write\
    .mode('overwrite')\
        .format('delta')\
            .option('mergeSchema', 'true')\
                .save(f'{bronze_path}/itens_pedido')
                
df_pedidos.write\
    .mode('overwrite')\
        .format('delta')\
            .option('mergeSchema', 'true')\
                .save(f'{bronze_path}/pedidos')
                
            
df_produtos.write\
    .mode('overwrite')\
        .format('delta')\
            .option('mergeSchema', 'true')\
                .save(f'{bronze_path}/produtos')
                
df_vendedores.write\
    .mode('overwrite')\
        .format('delta')\
            .option('mergeSchema', 'true')\
                .save(f'{bronze_path}/vendedores')
```
